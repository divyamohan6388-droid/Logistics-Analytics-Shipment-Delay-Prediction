# Business Recommendations: Logistics Shipment Delay Analysis

## 1. Project Objective

The objective of this project is to analyse shipment performance, identify patterns associated with late-delivery risk, and evaluate machine learning models for classifying the dataset's `Late_delivery_risk` label.

## 2. Key Business Findings

- **High delivery-risk prevalence:** 54.83% of records are labelled as having late-delivery risk.
- **Schedule overruns:** 57.28% of records have an actual shipping duration greater than the scheduled duration.
- **Shipping-mode differences:** Second Class has an average actual-minus-scheduled duration difference of 1.99 days. First Class records show a 100% schedule-overrun rate in this dataset, which warrants investigation.
- **Statistical evidence:** The analysis found a strong association between shipping mode and the late-delivery-risk label. This association does not establish that shipping mode itself causes delays.
- **Model performance:** Logistic Regression achieved an F1-score of 65.54% and ROC-AUC of 0.7370. Random Forest achieved an F1-score of 66.30% and ROC-AUC of 0.7369.

## 3. Recommended Actions

### A. Review Shipping-Mode Performance
Investigate why First Class and Second Class records show substantial schedule overruns. Compare promised delivery windows, carrier performance, routes, and operational constraints before changing service policies.

### B. Monitor Delivery KPIs
Build a recurring dashboard to track schedule-overrun rate, late-delivery-risk rate, average schedule deviation, and performance by shipping mode, region, and product category.

### C. Investigate Changes in Order Volume
Review the observed decline in record counts in late 2017. The available analysis identifies the change but does not establish its cause.

### D. Use Model Predictions Cautiously
The Random Forest model has a slightly higher recall and F1-score, while Logistic Regression has higher precision. Further threshold analysis and validation are needed before selecting a model for operational use.

### E. Improve Future Model Validation
Confirm which features are available at the actual prediction time. Exclude post-shipment information, validate against future periods, and evaluate false negatives carefully before using predictions to prioritise shipments.

## 4. Limitations

- The dataset contains order-item records, which may not correspond one-to-one with physical shipments.
- The target label is related to delivery outcome information, so its meaning and timing must be considered when interpreting model results.
- The models have only been evaluated on this dataset and have not been validated in a live logistics operation.
- The observed patterns are associations and should not automatically be interpreted as causal relationships.

## 5. Conclusion

The analysis reveals substantial schedule overruns and meaningful differences across shipping modes. The machine learning models provide a preliminary baseline for classifying the dataset's late-delivery-risk label. Further validation, operational investigation, and monitoring are required before deployment.
