```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
df = pd.read_csv(
    "../data/raw/DataCoSupplyChainDataset.csv",
    encoding="latin-1",
    sep=","
)

df.head()
```

```python
# checking for null values
missing = df.isnull().sum()

missing[missing > 0].sort_values(ascending=False)
```

```python
# duplicate rows
df.duplicated().sum()
```

```python
# seeing targer variable
df["Late_delivery_risk"].value_counts()
df["Late_delivery_risk"].value_counts(normalize=True) * 100
```

```python
df["Delivery Status"].value_counts()
```

```python
# checking missing values properly
missing_summary = pd.DataFrame({
    "Missing Values": missing[missing > 0].sort_values(ascending=False),
    "Percentage": (missing[missing > 0].sort_values(ascending=False) / len(df)) * 100
})
missing_summary
```

```python
# duplicate rows
duplicate_count = df.duplicated().sum()

print(f"Duplicate rows: {duplicate_count}")
print(f"Duplicate percentage: {duplicate_count / len(df) * 100:.2f}%")
```

```python
# checking categorical columns
categorical_columns = df.select_dtypes(include=["object"]).columns

for col in categorical_columns:
    print(f"Column: {col}")
    print(df[col].nunique(), "unique values")
    print(df[col].value_counts(dropna=False).head(10))
```

```python
# checking numerical columns
numerical_columns = df.select_dtypes(include=[np.number]).columns
df[numerical_columns].describe().T
```

```python
# understanding dates
date_candidates = [
    col for col in df.columns if "date" in col.lower() or "time" in col.lower()
]

date_candidates
```

```python
for col in date_candidates:
    print(f"Column: {col}")
    print(df[col].nunique(), "unique values")
    print(df[col].value_counts(dropna=False).head(10))
```
