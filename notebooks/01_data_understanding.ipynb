### Prediction Leakage Note

`Delivery Status`, `Days for shipping (real)`, and the derived `shipping_delay_days` contain information that becomes available after or during the shipment outcome. These variables will therefore not be used as predictors in the primary early-stage shipment-delay prediction model.

They may still be used for retrospective logistics performance analysis and KPI development.

```python
# Detect potential outliers using the IQR method

columns_to_check = [
    "Benefit per order",
    "Sales per customer",
    "Order Item Product Price",
    "Order Item Discount",
    "Order Item Total",
    "shipping_delay_days"
]

outlier_results = []

for col in columns_to_check:
    q1 = df_clean[col].quantile(0.25)
    q3 = df_clean[col].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (df_clean[col] < lower_bound)
        | (df_clean[col] > upper_bound)
    )

    outlier_results.append({
        "Column": col,
        "Q1": round(q1, 2),
        "Q3": round(q3, 2),
        "IQR": round(iqr, 2),
        "Lower Bound": round(lower_bound, 2),
        "Upper Bound": round(upper_bound, 2),
        "Potential Outliers": int(outlier_mask.sum()),
        "Outlier Percentage": round(
            outlier_mask.mean() * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary)
```

```python
# Count each observed shipping-delay value

delay_distribution = (
    df_clean["shipping_delay_days"]
    .value_counts()
    .sort_index()
    .rename_axis("Shipping Delay (Days)")
    .reset_index(name="Number of Records")
)

delay_distribution["Percentage"] = (
    delay_distribution["Number of Records"]
    / len(df_clean) * 100
).round(2)

display(delay_distribution)
```

```python
# Validate the shipping-delay distribution

print("Total records:", delay_distribution["Number of Records"].sum())
print("Dataset records:", len(df_clean))

print(
    "Percentages total:",
    round(delay_distribution["Percentage"].sum(), 2),
    "%"
)
```
